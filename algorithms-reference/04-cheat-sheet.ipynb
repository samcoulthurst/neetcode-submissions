## Gotchas that actually cost marks

**Binary search — the two shapes are not interchangeable.** Both live in 02 §1. Pick one and commit:

| | Exact match | Lower bound |
|---|---|---|
| Loop | `while l <= r` | `while l < r` |
| Initial `r` | `len(nums) - 1` | `len(nums)` |
| Move right | `r = m - 1` | `r = m` |
| Returns | inside the loop | `l`, after the loop |

Using `l <= r` with `r = m` is an **infinite loop**. This is the single most common bug
across these notebooks.

**Linked lists**
- Save `curr.next` before you overwrite it, or the rest of the list is gone.
- Use a dummy head whenever the head might change — it deletes your special cases.
- `while fast and fast.next` — both checks, because you dereference two levels deep.

**Trees**
- Snapshot `len(queue)` before the inner BFS loop or your levels merge.
- Validate a BST with inherited bounds, never a parent-only comparison.
- Recurse once per child and store the result; recursing twice is exponential.

**Sliding window**
- `while` to shrink, not `if` — one addition can force several removals.
- *Longest* records after the shrink loop; *shortest* records inside it.
- `del` zero-count keys if your validity test reads `len(count)`.

**General**
- Never mutate a list while iterating over it — iterate a copy, or build a new one.
- `list.pop(0)` is O(n). `deque.popleft()` is O(1).
- Check the empty input and the single-element input before you say you're done.
- Integer division `//` floors toward negative infinity: `-7 // 2 == -4`, not `-3`.

## Python idioms worth having in your fingers

Small things that save time and read well under pressure.

```python
from collections import Counter, defaultdict, deque

nums = [3, 1, 4, 1, 5]

print(sorted(nums, reverse=True))            # [5, 4, 3, 1, 1]
print(sorted(["bb", "a"], key=len))          # sort by a computed key
print(list(enumerate("ab")))                 # [(0, 'a'), (1, 'b')] -- index AND value
print(list(zip([1, 2], "ab")))               # [(1, 'a'), (2, 'b')] -- parallel iteration
print(list(range(len(nums) - 1, -1, -1)))    # [4, 3, 2, 1, 0] -- backwards

print(Counter("aab").most_common(1))         # [('a', 2)]
print("".join(reversed("abc")))              # 'cba'
print([x for x in nums if x > 2])            # [3, 4, 5] -- filter
print(float("inf"), float("-inf"))           # sentinels for min/max accumulators

# Tuple unpacking: swap and multi-assign without a temp variable
l, r = 0, len(nums) - 1
l, r = r, l
print(l, r)                                  # 4 0

# divmod: flat index -> 2D coordinates (see 02 §1)
print(divmod(7, 3))                          # (2, 1)
```
