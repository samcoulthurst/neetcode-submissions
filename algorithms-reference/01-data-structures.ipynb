## 1. Linked list

**Complexity:** traversal O(n) time, O(1) space.

The dummy head exists so you never write a special case for "is this the first
node?". Build onto `dummy`, return `dummy.next`. Forgetting `.next` on the return is an
easy mistake.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
def traverse(head):
    """Walk a list, collecting values. The shape of every linked-list loop."""
    # O(n) time, O(1) space.
    out = []
    curr = head
    while curr:
        out.append(curr.val)
        curr = curr.next          # ALWAYS advance, or you loop forever
    return out


def build_with_dummy(vals):
    # O(n) time, O(1) space.
    """The dummy-head idiom: a throwaway node so appending is uniform."""
    dummy = ListNode()            # placeholder, never part of the answer
    curr = dummy
    for v in vals:
        curr.next = ListNode(v)
        curr = curr.next
    return dummy.next             # skip the placeholder
```

```python
def reverse_list(head):
    # O(n) time, O(1) space.
    prev = None
    curr = head
    while curr:
        nxt = curr.next           # 1. save the rest of the list
        curr.next = prev          # 2. flip this pointer backwards
        
        prev = curr               # 3. shuffle both pointers forward
        curr = nxt
    return prev                   # curr is None here; prev is the new head
```

```python
def merge_two_lists(l1, l2):
    # O(n + m) time, O(1) space
    dummy = ListNode()
    curr = dummy
    while l1 and l2:              
        if l1.val <= l2.val:
            curr.next = l1
            l1 = l1.next
        else:
            curr.next = l2
            l2 = l2.next
        curr = curr.next
    curr.next = l1 or l2          # attach whichever tail remains
    return dummy.next
```

## 2. Binary tree

**Complexity:** `build_tree` is O(n).

The level-order list format uses `None` for missing children, and those `None`s are *not*
enqueued — only real nodes get children assigned to them.

```python
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def build_tree(vals):
    """LeetCode level-order format, None for missing. [1,2,3,None,4] -> root."""
    # O(n) time, O(n) space.
    if not vals:
        return None
    root = TreeNode(vals[0])
    queue = deque([root])
    i = 1
    while queue and i < len(vals):
        node = queue.popleft()
        if i < len(vals) and vals[i] is not None:
            node.left = TreeNode(vals[i])
            queue.append(node.left)   # only real nodes get enqueued
        i += 1
        if i < len(vals) and vals[i] is not None:
            node.right = TreeNode(vals[i])
            queue.append(node.right)
        i += 1
    return root
```

```python
#         1
#        / \
#       2   3
#      / \
#     4   5
root = build_tree([1, 2, 3, 4, 5])
```

## 3. Stack, queue, and counting

**Complexity:** all operations shown are O(1).

Never use `list.pop(0)` for a queue — it's O(n) and will time you out. Use `deque.popleft()`.
`Counter` and `defaultdict` save you the "is this key present yet?" branch every time.

```python
from collections import deque, defaultdict, Counter
# --- Stack: a plain list. Append/pop from the RIGHT end. ---
stack = []
stack.append(1)
stack.append(2)
print(stack[-1], stack.pop(), stack)        # 2 2 [1]   (peek, pop, remainder)

# --- Queue: deque. Append right, pop LEFT. ---
queue = deque([1, 2, 3])
queue.append(4)
print(queue.popleft(), list(queue))         # 1 [2, 3, 4]

# --- Counting: Counter for tallies, defaultdict for grouping ---
counts = Counter("aabbbc")
print(counts, counts["b"], counts["z"])     # missing key -> 0, no KeyError

groups = defaultdict(list)                  # values default to []
for word in ["eat", "tea", "tan"]:
    groups[tuple(sorted(word))].append(word)
print(dict(groups))                         # anagrams grouped by sorted-letter key
```
